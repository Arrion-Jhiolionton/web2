```python
# ===========================
# 🔹 YOLO Inference from Local Model & Local Image
# ===========================

# 1️⃣ Install ultralytics if not already
!pip install ultralytics --quiet

# 2️⃣ Import necessary modules
from ultralytics import YOLO
from google.colab import files
from IPython.display import display
from PIL import Image
import os

# 3️⃣ Upload the YOLO model (.pt) from local machine
print("⬆️ Upload your YOLO model file (.pt)")
uploaded_model = files.upload()  # choose prescription_yolo_best.pt
model_path = list(uploaded_model.keys())[0]
print(f"✅ Model uploaded: {model_path}")

# Load the model
model = YOLO(model_path)
print("✅ Model loaded successfully!")
```

```python
# 4️⃣ Upload one or more new images to test
print("⬆️ Upload images to test")
uploaded_images = files.upload()
image_paths = list(uploaded_images.keys())
print(f"✅ Uploaded images: {image_paths}")
```

```python
# 5️⃣ Output folder in Colab (temporary, or you can download later)
output_dir = 'yolo_outputs'
os.makedirs(output_dir, exist_ok=True)

# 6️⃣ Run prediction and save results (fixed)
import os

# Predict, show, and save
for img_name in image_paths:
    results_list = model.predict(img_name)

    for results in results_list:
        results.show()

        # Create output path with same filename
        save_path = os.path.join(output_dir, img_name)
        results.save(save_path)
        print(f"✅ Results saved to {save_path}")
```
